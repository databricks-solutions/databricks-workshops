```python
from databricks.sdk import WorkspaceClient
from databricks.sdk.service.jobs import NotebookTask, Source, Task, TaskDependency
```

```python
w = WorkspaceClient()

setup_notebook_path = f"/Workspace/Users/{w.current_user.me().user_name}/databricks-workshops/streaming_etl/0_setup/setup"
notebook_path = f"/Workspace/Users/{w.current_user.me().user_name}/databricks-workshops/streaming_etl/1_structured_streaming/0_flights_bootstrap"

created_job = w.jobs.create(
  name = "flights_bootstrap",
  tasks = [
    Task(
      notebook_task = NotebookTask(
      notebook_path = setup_notebook_path,
      source = Source("WORKSPACE")
      ),
      task_key = "Run_0_setup",
   ),
    Task(
      notebook_task = NotebookTask(
      notebook_path = notebook_path,
      source = Source("WORKSPACE")
      ),
      task_key = "Run_0_flights_bootstrap",
      depends_on = [TaskDependency("Run_0_setup")]
   )
  ]
)

print(created_job.job_id)
```

```python
job_run = w.jobs.run_now(created_job.job_id)
```
